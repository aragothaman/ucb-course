```python
  import plotly.express as px
  import pandas as pd
```

```python
df = pd.read_csv('data/world-happiness-report/2019.csv')
df.head()

df['Country'] = df['Country or region']

df.head()

```

```python
  fig = px.choropleth(
      df,
      locations='Country',          # column with country names
      locationmode='country names', # or 'ISO-3' if you have country codes
      color='Score',
      hover_name='Country',
      color_continuous_scale='RdYlGn',  # red=low, green=high
      title='World Happiness Score 2023'
  )
  fig.show()
```
